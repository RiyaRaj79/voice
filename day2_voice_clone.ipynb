```python
from google.colab import files
uploaded = files.upload()
voice_sample = list(uploaded.keys())[0]
```

```python
from f5_tts.api import F5TTS
f5tts = F5TTS()
```

```python
from IPython.display import Audio

english_text = "My name is Riya. I live here. I don't know where to go. Please help me."
sample_text = "I agree to let Voice Translator use my voice to speak my translations. This is my real voice."

f5tts.infer(
    ref_file=voice_sample,
    ref_text=sample_text,
    gen_text=english_text,
    file_wave="cloned.wav",
)
Audio("cloned.wav")
```

```python
from faster_whisper import WhisperModel
whisper_model = WhisperModel("medium", device="cuda", compute_type="float16")
```

```python
from google.colab import files
uploaded = files.upload()
hindi_audio = list(uploaded.keys())[0]
```

```python
from IPython.display import Audio

# 1. Hindi speech -> English text
segments, info = whisper_model.transcribe(hindi_audio, task="translate")
english_text = " ".join(s.text for s in segments)
print("Detected language:", info.language)
print("English:", english_text)

# 2. English text -> speech in your voice
f5tts.infer(
    ref_file=voice_sample,
    ref_text=sample_text,
    gen_text=english_text,
    file_wave="final_output.wav",
)
Audio("final_output.wav")
```
